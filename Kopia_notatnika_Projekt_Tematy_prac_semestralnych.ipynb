#### Wstęp teoretyczny: trzy rodzaje pamięci masowej

| | Pamięć **blokowa** | Pamięć **plikowa** | Pamięć **obiektowa** |
|---|---|---|---|
| Jak wygląda dla użytkownika | „goły dysk”, który trzeba sformatować | katalogi i pliki, jak na pendrivie | płaska przestrzeń **obiektów** w **zasobnikach** (*buckets*) |
| Jak się z nią rozmawia | system operacyjny maszyny | protokoły sieciowe NFS, SMB | **HTTP** — każdy obiekt ma adres |
| Typowe zastosowanie | dysk systemowy maszyny wirtualnej, baza danych | wspólny katalog dla wielu serwerów | zdjęcia, kopie zapasowe, archiwa, dane do analizy, strony statyczne |
| AWS / Azure / Google Cloud | EBS / Managed Disks / Persistent Disk | EFS / Azure Files / Filestore | **S3** / Blob Storage / Cloud Storage |

**Obiekt** = dane (dowolne bajty) + **klucz** (nazwa, np. `zdjecia/2026/kot.jpg`) + **metadane** (np. typ zawartości, autor). Ukośniki w kluczu tylko *udają* katalogi: naprawdę są to **prefiksy** nazw, a „otwarcie katalogu” to wyszukanie kluczy o danym prefiksie.

Cechy, dla których pamięć obiektowa jest dziś podstawą chmury:

- **ogromna trwałość** — dane przechowywane są w wielu kopiach w kilku strefach dostępności; dostawcy deklarują trwałość rzędu 99,999999999% w skali roku;
- **praktycznie nieograniczona pojemność** i płatność za faktycznie zajęte gigabajty;
- **wersjonowanie** — nadpisanie lub usunięcie obiektu nie niszczy poprzednich wersji;
- **klasy przechowywania** — dane „gorące” kosztują więcej za przechowywanie, „zimne” (archiwalne) mniej, ale drożej i wolniej się je odczytuje;
- **prywatność domyślna** — nowy zasobnik jest prywatny; dostęp czasowy daje się przez **adres podpisany** (*presigned URL*).

### Zadanie 1.1. Emulator i pierwszy zasobnik

Komórka instaluje bibliotekę i emulator, uruchamia go w tle i tworzy pierwszy zasobnik. Klucze dostępu `test`/`test` są **fikcyjne** — emulator ich nie sprawdza, ale biblioteka boto3 wymaga, by jakieś były podane.

````python
!pip -q install "moto[server]" boto3

import logging, boto3
from moto.server import ThreadedMotoServer
logging.getLogger("werkzeug").setLevel(logging.ERROR)     # wyciszenie dziennika emulatora

ADRES = "http://127.0.0.1:5050"
REGION = "eu-central-1"                                    # Frankfurt — region, nie miasto w kodzie
import socket
with socket.socket() as gniazdo:
    juz_dziala = gniazdo.connect_ex(("127.0.0.1", 5050)) == 0
if juz_dziala:
    print("Emulator już działa — nie uruchamiam go drugi raz.")
else:
    serwer = ThreadedMotoServer(ip_address="127.0.0.1", port=5050); serwer.start()

dostep = dict(endpoint_url=ADRES, region_name=REGION, aws_access_key_id="test", aws_secret_access_key="test")
s3 = boto3.client("s3", **dostep)

ZASOBNIK = "archiwum-tu-wpisz-imie"                       # tylko małe litery, cyfry i myślniki!
s3.create_bucket(Bucket=ZASOBNIK, CreateBucketConfiguration={"LocationConstraint": REGION})
print("Zasobniki na koncie:", [z["Name"] for z in s3.list_buckets()["Buckets"]])
print("Region zasobnika:  ", s3.get_bucket_location(Bucket=ZASOBNIK)["LocationConstraint"])
````

### Zadanie 1.2. Obiekty, klucze i metadane

````python
s3.put_object(Bucket=ZASOBNIK, Key="notatki/powitanie.txt", Body="Dzień dobry".encode("utf-8"),
              ContentType="text/plain; charset=utf-8", Metadata={"autor": "tu-wpisz-imie", "kategoria": "notatka"})
s3.put_object(Bucket=ZASOBNIK, Key="zdjecia/2026/kot.jpg", Body=b"\xff" * 2048, ContentType="image/jpeg")
s3.put_object(Bucket=ZASOBNIK, Key="zdjecia/2026/pies.jpg", Body=b"\xff" * 3072, ContentType="image/jpeg")

print("=== Wszystkie obiekty ===")
for o in s3.list_objects_v2(Bucket=ZASOBNIK)["Contents"]:
    print(f"{o['Key']:<28}{o['Size']:>6} B")

print("\n=== Tylko prefiks zdjecia/ ===")
for o in s3.list_objects_v2(Bucket=ZASOBNIK, Prefix="zdjecia/").get("Contents", []):
    print(o["Key"])

naglowek = s3.head_object(Bucket=ZASOBNIK, Key="notatki/powitanie.txt")
print("\nMetadane:", naglowek["Metadata"])
print("Rozmiar: ", naglowek["ContentLength"], "B")
print("ETag:    ", naglowek["ETag"])
````

#### Wstęp teoretyczny: obraz, kontener, rejestr

Na wykładzie 1 porównaliśmy maszyny wirtualne z kontenerami. Teraz potrzebujemy czterech pojęć:

| Pojęcie | Czym jest | Analogia |
|---|---|---|
| **Obraz** (*image*) | niezmienny „szablon”: system plików z aplikacją i wszystkim, czego potrzebuje | przepis kulinarny wraz ze składnikami w słoikach |

| **Plik `Dockerfile`** | lista kroków budowy obrazu: od jakiego obrazu zacząć, co skopiować, co uruchomić | treść przepisu |
| **Rejestr** (*registry*) | serwer przechowujący obrazy, z którego pobiera je każdy komputer na świecie | księgarnia z przepisami |

Obraz składa się z **warstw** — każdy krok `Dockerfile` dodaje jedną. Warstwy są współdzielone: dziesięć obrazów opartych na `nginx:alpine` przechowuje wspólną podstawę tylko raz. Obraz identyfikuje się nazwą i **znacznikiem** (*tag*), np. `moja-strona:1.0`.

#### Wstęp teoretyczny: po co orkiestracja?

Jeden kontener uruchamia się poleceniem `docker run`. Ale co, gdy kontenerów jest sto, na dziesięciu serwerach, a o trzeciej w nocy jeden serwer ulega awarii? Ktoś musi ponownie uruchomić kontenery w innym miejscu, rozłożyć ruch, wymienić wersje aplikacji bez wyłączania usługi. Tym „kimś” jest **orkiestrator** — dziś niemal zawsze **Kubernetes**.

Najważniejsza idea Kubernetesa to **stan pożądany**. Nie wydaje się poleceń w rodzaju „uruchom kontener”, lecz **deklaruje się**: „mają działać 3 kopie tej aplikacji”. Kubernetes w pętli porównuje stan pożądany z rzeczywistym i usuwa różnice — to **pętla uzgadniania** (*reconciliation loop*).

| Obiekt | Rola |
|---|---|
| **Węzeł** (*node*) | maszyna (fizyczna lub wirtualna), na której działają kontenery |
| **Pod** | najmniejsza jednostka: jeden lub kilka kontenerów zawsze uruchamianych razem |
| **Wdrożenie** (*Deployment*) | deklaracja „ile kopii którego obrazu”; zarządza aktualizacjami |
| **Zbiór replik** (*ReplicaSet*) | pilnuje, żeby liczba podów zgadzała się z deklaracją |
| **Usługa** (*Service*) | stały adres i nazwa dla zmiennego zbioru podów; równoważy ruch |
| **Mapa konfiguracji** / **sekret** | konfiguracja i dane poufne przekazywane do podów |

#### Wstęp teoretyczny: Kubernetes w Dockerze

W chmurze klaster to zwykle usługa zarządzana: **EKS** (AWS), **AKS** (Azure), **GKE** (Google Cloud) — dostawca utrzymuje warstwę sterującą, klient płaci głównie za węzły. Do nauki i testów używa się klastrów lokalnych. Narzędzie **kind** (*Kubernetes IN Docker*) uruchamia każdy węzeł klastra jako… kontener Dockera. Kontener udaje maszynę, na której działają kolejne kontenery — wirtualizacja wewnątrz wirtualizacji, wewnątrz Codespace, wewnątrz maszyny wirtualnej w centrum danych Microsoftu.

Klastrem steruje się narzędziem **`kubectl`**, a w projekcie — plikami **manifestów** w formacie YAML, przechowywanymi w repozytorium. Manifest to właśnie zapisany stan pożądany.

#### Wstęp teoretyczny: przetwarzanie bezserwerowe i zdarzenia

**Przetwarzanie bezserwerowe** (*serverless*) oznacza, że klient dostarcza **kod** i mówi, **kiedy** ma się wykonać, a dostawca sam przydziela maszynę, uruchamia kod i maszynę zwalnia. Płaci się za faktyczny czas wykonania; gdy nic się nie dzieje, koszt wynosi zero.

Kod uruchamiają **zdarzenia** (*triggers*):

| Rodzaj zdarzenia | Przykład w chmurze | Przykład w GitHub Actions |
|---|---|---|
| zmiana w danych | nowy plik w zasobniku S3 | `push` — nowe zatwierdzenie w repozytorium |
| harmonogram | Amazon EventBridge Scheduler, Cloud Scheduler | `schedule` z wyrażeniem **cron** |
| żądanie HTTP | wywołanie funkcji przez adres | `workflow_dispatch` — uruchomienie ręczne lub przez API |

**GitHub Actions** jest usługą zdarzeniową tego samego rodzaju: każde uruchomienie dostaje **świeżą maszynę wirtualną** (*runner*), która po zakończeniu znika. Dla repozytoriów publicznych standardowe maszyny są bezpłatne; w prywatnych zużywa się limit minut, a każde zadanie zaokrągla się w górę do pełnej minuty — typowa cecha usług mierzonych.

#### Wstęp teoretyczny: potok danych i jego pułapki

Potok **ETL** (*extract, transform, load*) to trzy kroki: **pobierz** dane ze źródła, **przekształć** je do potrzebnej postaci, **zapisz** w miejscu docelowym. U nas: interfejs NBP → wiersz CSV → plik w repozytorium → strona w GitHub Pages.

**Źródło danych.** NBP udostępnia bezpłatny interfejs bez rejestracji i kluczy, np.:

`https://api.nbp.pl/api/exchangerates/rates/a/eur/last/10/?format=json` — 10 ostatnich średnich kursów euro z tabeli A.

Tabela A publikowana jest w dni robocze; w weekendy i święta nowej tabeli nie ma.

**Trzy pułapki każdego potoku uruchamianego automatycznie:**

1. **Idempotentność** — dwukrotne uruchomienie z tymi samymi danymi nie może psuć wyniku (np. dublować wierszy).
2. **Czas** — harmonogramy w chmurze podaje się zwykle w **UTC**. Dziś, 25 października 2026 r., Polska przeszła na czas zimowy (UTC+1); latem obowiązuje UTC+2. Ponadto uruchomienia z harmonogramu w GitHub Actions bywają opóźnione, a w repozytoriach publicznych bez aktywności przez 60 dni harmonogram zostaje wyłączony.
3. **Brak danych** — źródło może nie odpowiedzieć albo nie mieć nowych danych (weekend). Potok ma wtedy zakończyć się spokojnie, a nie zapisać śmieci.

**Najmniejsze uprawnienia:** potok dostaje tylko te uprawnienia, których potrzebuje (u nas: zapis do zawartości repozytorium) — przypomnienie z modelu odpowiedzialności współdzielonej.

### Zadanie 4.3. Harmonogram w UTC

Harmonogram zapisuje się wyrażeniem **cron** złożonym z pięciu pól:

```
┌──────── minuta (0–59)
│ ┌────── godzina (0–23, w UTC!)
│ │ ┌──── dzień miesiąca (1–31)
│ │ │ ┌── miesiąc (1–12)
│ │ │ │ ┌ dzień tygodnia (0–6, 0 = niedziela)
* * * * *
```

`*` oznacza „każdy”, `*/15` — „co 15”, `1-5` — „od 1 do 5”, `1,15` — „1 i 15”. Przykład: `30 6 * * 1` to każdy poniedziałek o 6:30 UTC.

Proszę wpisać wyrażenia w komórce poniżej i ją uruchomić — sprawdza się sama.

````python
import hashlib

def _skrot(pytanie, odpowiedz):
    return hashlib.sha256(f"{pytanie}|{' '.join(odpowiedz.split()).lower()}".encode()).hexdigest()[:12]

def sprawdz(odpowiedzi, wzorzec):
    dobrze = 0
    for pytanie, odp in odpowiedzi.items():
        if not odp.strip():
            znak = "⬜ brak odpowiedzi"
        elif _skrot(pytanie, odp) in wzorzec[pytanie]:
            znak, dobrze = "✅", dobrze + 1
        else:
            znak = "❌ proszę spróbować jeszcze raz"
        print(f"{pytanie:<60}{odp:<18}{znak}")
    print(f"\nWynik: {dobrze} / {len(wzorzec)}")

odpowiedzi_cron = {
    "Co 15 minut": "*/15 * * * *",
    "Codziennie o północy UTC": "0 0 * * *",
    "W dni robocze (pon.–pt.) o 12:20 UTC": "20 12 * * 1-5",
    "W każdą niedzielę o 6:00 UTC": "0 6 * * 0",
    "Pierwszego dnia każdego miesiąca o 8:30 UTC": "30 8 1 * *",
}

wzorzec_cron = {
    'Co 15 minut': ['bfd2648ea590', 'c9f1fed4bdda', 'fcac24576995'],
    'Codziennie o północy UTC': ['b8d2516de092'],
    'W dni robocze (pon.–pt.) o 12:20 UTC': ['08a42a5263a5', '12761fa41f9d', 'bf9415545fcc'],
    'W każdą niedzielę o 6:00 UTC': ['71989f979aeb', '9a0c24fdb111', 'c097ff043c4f'],
    'Pierwszego dnia każdego miesiąca o 8:30 UTC': ['b38b854f7cf7'],
}
sprawdz(odpowiedzi_cron, wzorzec_cron)
````

#### Wstęp teoretyczny: dlaczego infrastruktura jako kod?

Infrastrukturę można „wyklikać” w konsoli dostawcy. Działa — do pierwszej pomyłki, do drugiego środowiska, do pytania „kto i kiedy to zmienił?”. **Infrastruktura jako kod** (*Infrastructure as Code*, IaC) oznacza, że serwery, sieci, bazy i uprawnienia opisuje się **w plikach tekstowych**, przechowuje w repozytorium i tworzy automatycznie.

| Bez IaC | Z IaC |
|---|---|
| instrukcja „kliknij tu, potem tam” | plik, który wykonuje się sam |
| dwa środowiska nigdy nie są identyczne | `dev` i `prod` różnią się tylko wartościami zmiennych |
| historia zmian w pamięci administratora | historia zmian w repozytorium, z przeglądem przed wdrożeniem |
| odtworzenie po katastrofie: dni | odtworzenie po katastrofie: minuty |

Opis jest **deklaratywny**: pisze się, *co* ma istnieć, a nie *jak* to zbudować. Narzędzie samo wylicza różnicę między stanem obecnym a pożądanym i proponuje **plan** zmian.

#### Wstęp teoretyczny: OpenTofu, dostawcy, stan

**OpenTofu** to otwartoźródłowa odmiana (*fork*) narzędzia Terraform, rozwijana pod opieką Linux Foundation od 2023 r., gdy Terraform przeszedł na licencję o ograniczonej otwartości. Język (HCL) i sposób pracy są w obu narzędziach praktycznie takie same.

| Pojęcie | Znaczenie |
|---|---|
| **Dostawca** (*provider*) | wtyczka umiejąca rozmawiać z konkretną platformą: AWS, Azure, Google Cloud… albo **Docker** |
| **Zasób** (*resource*) | jeden element infrastruktury, np. kontener, sieć, maszyna wirtualna |
| **Zmienna** / **wynik** (*variable* / *output*) | parametry wejściowe i informacje zwracane po wdrożeniu |
| **Stan** (*state*) | plik, w którym narzędzie pamięta, co już utworzyło — **może zawierać sekrety**, więc nie trafia do publicznego repozytorium |
| **Dryf** (*drift*) | rozjazd między stanem zapisanym a rzeczywistością, np. po ręcznej zmianie |

Cykl pracy jest zawsze ten sam: **`init`** (pobranie dostawców) → **`plan`** (co się zmieni?) → **`apply`** (zmień) → … → **`destroy`** (posprzątaj).

W projekcie dostawcą będzie **Docker** w Codespace — zamiast maszyn wirtualnych AWS powstają kontenery, ale język, plan, stan i dryf są dokładnie takie same. Przy dostawcy chmurowym zmieniłyby się nazwy zasobów, nie sposób myślenia.

#### Wstęp teoretyczny: całkowity koszt posiadania

Porównanie „serwer kosztuje 15 000 zł, a maszyna w chmurze 300 zł miesięcznie” jest zawsze fałszywe. Porównuje się **całkowity koszt posiadania** (*Total Cost of Ownership*, TCO) w tym samym horyzoncie, np. 3 lat:

| Własna serwerownia | Chmura |
|---|---|
| zakup sprzętu i licencji (z góry — **CapEx**) | opłaty za użycie (co miesiąc — **OpEx**) |
| prąd, chłodzenie, pomieszczenie, łącze | moc obliczeniowa, pamięć masowa, **ruch wychodzący** |
| serwis, wymiana dysków, gwarancja | plan wsparcia technicznego |
| praca administratora | praca administratora (mniej, ale nie zero!) |
| zapas mocy „na szczyt” | elastyczność — ale też ryzyko zasobów zapomnianych |

**FinOps** to praktyka zarządzania kosztami chmury w trzech krokach: **widoczność** (kto i na co wydaje — stąd **znaczniki** *tags* na zasobach), **optymalizacja** (wyłączanie, zmniejszanie, rabaty za zobowiązanie), **działanie** (budżety, alarmy, odpowiedzialność zespołów).

#### Wstęp teoretyczny: strategie migracji i miejsce danych

Każdy system firmy przy migracji dostaje jedną z sześciu decyzji (tzw. **6R**):

| Strategia | Znaczenie | Przykład |
|---|---|---|
| **Rehost** — „przenieś jak jest” | maszyna fizyczna → maszyna wirtualna w chmurze | serwer aplikacji magazynowej |
| **Replatform** — „przenieś z poprawkami” | np. własna baza → baza zarządzana | PostgreSQL → Cloud SQL / RDS |
| **Refactor** — „przebuduj” | przepisanie na kontenery lub funkcje | sklep internetowy na mikrousługach |
| **Repurchase** — „kup gotowe” | zastąpienie usługą SaaS | serwer poczty → Microsoft 365 / Google Workspace |
| **Retire** — „wyłącz” | system nikomu niepotrzebny | stary serwer raportów |
| **Retain** — „zostaw” | na razie bez zmian (prawo, koszty, ryzyko) | system kadrowo-płacowy |

**Gdzie mogą leżeć dane?** Firma z Unii Europejskiej przetwarzająca dane osobowe podlega **RODO**. Wybór regionu w UE jest warunkiem rozsądnym, ale **niewystarczającym**: potrzebna jest też umowa powierzenia przetwarzania z dostawcą, szyfrowanie, kontrola dostępu, rejestr czynności przetwarzania — znów model odpowiedzialności współdzielonej. Niektóre branże (bankowość, ochrona zdrowia, administracja) mają wymagania dodatkowe.

### Zadanie 6.3. Model TCO: własna serwerownia czy chmura?

Firma potrzebuje 4 serwerów aplikacji. Wszystkie parametry są umowne i można je zmieniać. Proszę uruchomić komórkę, a następnie — zmieniając parametry — odpowiedzieć na pytania.

````python
import numpy as np, matplotlib.pyplot as plt

MIESIECY = 36

# --- Własna serwerownia (USD) ---
ZAKUP_SERWEROW    = 16000     # zakup sprzętu i licencji — z góry
WDROZENIE         = 3000      # montaż, okablowanie, konfiguracja — z góry
PRAD_I_CHLODZENIE = 150       # miesięcznie
ADMIN_WLASNA      = 500       # miesięcznie: część etatu administratora
NAPRAWY_ROCZNIE   = 800       # co 12 miesięcy: wymiany dysków, serwis

# --- Chmura (USD) ---
LICZBA_VM      = 4
VM_ZA_GODZINE  = 0.40         # cena jednej maszyny na żądanie
GODZIN_PRACY   = 730          # godzin w miesiącu (730 = całą dobę)
DYSK_GB, CENA_GB_MIES = 500, 0.08
WYSYLKA_GB, CENA_WYSYLKI_GB = 200, 0.09
ADMIN_CHMURA   = 300          # miesięcznie: mniej pracy, ale nie zero
RABAT_ZOBOWIAZANIE = 0.40     # rabat za 3-letnie zobowiązanie

m = np.arange(1, MIESIECY + 1)
wlasna = ZAKUP_SERWEROW + WDROZENIE + m * (PRAD_I_CHLODZENIE + ADMIN_WLASNA) + (m // 12) * NAPRAWY_ROCZNIE

def miesiac_chmury(rabat=0.0, godziny=GODZIN_PRACY):
    return (LICZBA_VM * VM_ZA_GODZINE * godziny * (1 - rabat)
            + DYSK_GB * CENA_GB_MIES + WYSYLKA_GB * CENA_WYSYLKI_GB + ADMIN_CHMURA)

chmura_na_zadanie = m * miesiac_chmury()
chmura_zobowiazanie = m * miesiac_chmury(RABAT_ZOBOWIAZANIE)

def kiedy_drozsza(chmura):
    """Pierwszy miesiąc, od którego narastający koszt chmury przekracza koszt własnej serwerowni."""
    i = np.where(chmura > wlasna)[0]
    return f"od miesiąca {m[i[0]]}" if i.size else "w ciągu 36 miesięcy — nigdy"

for n in (12, 24, 36):
    print(f"Po {n:>2} mies.: własna {wlasna[n-1]:>9,.0f}   chmura na żądanie {chmura_na_zadanie[n-1]:>9,.0f}"
          f"   chmura z zobowiązaniem {chmura_zobowiazanie[n-1]:>9,.0f}  USD")
print("\nChmura na żądanie droższa od własnej serwerowni:     ", kiedy_drozsza(chmura_na_zadanie))
print("Chmura z zobowiązaniem droższa od własnej serwerowni:", kiedy_drozsza(chmura_zobowiazanie))

fig, ax = plt.subplots(figsize=(11, 4.2))
ax.plot(m, wlasna, color="#b91c1c", lw=2.5, label="Własna serwerownia")
ax.plot(m, chmura_na_zadanie, color="#0b5cad", lw=2.5, label="Chmura — na żądanie")
ax.plot(m, chmura_zobowiazanie, color="#0f766e", lw=2.5, ls="--", label="Chmura — z zobowiązaniem")
ax.set_xlabel("Miesiąc"); ax.set_ylabel("Koszt narastająco [USD]")
ax.set_title("Całkowity koszt posiadania w horyzoncie 3 lat", fontweight="bold")
ax.legend(frameon=False); ax.spines[["top", "right"]].set_visible(False)
plt.show()
````
